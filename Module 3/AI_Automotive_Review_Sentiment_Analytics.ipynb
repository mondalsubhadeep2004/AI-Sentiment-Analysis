```python
# STEP 6: Clean sentiment data

import pandas as pd

# Keep only valid sentiment labels
valid_sentiments = ["pos", "neg", "neu"]

df = annotations[
    annotations["sentiment"].isin(valid_sentiments)
].copy()

# Remove rows with missing review/aspect
df = df.dropna(subset=["review", "aspect"])

# Remove empty text
df = df[
    (df["review"].astype(str).str.strip() != "") &
    (df["aspect"].astype(str).str.strip() != "")
]

# Convert labels to readable names
label_map = {
    "neg": "Negative",
    "neu": "Neutral",
    "pos": "Positive"
}

df["sentiment_label"] = df["sentiment"].map(label_map)

print("Original records:", len(annotations))
print("Clean records:", len(df))

print("\nRemoved records:", len(annotations) - len(df))

print("\nFinal sentiment distribution:")
print(df["sentiment_label"].value_counts())

display(df[["review", "aspect", "opinion", "sentiment_label"]].head(10))
```

```python
# STEP 7: Prepare BERT input

df["bert_text"] = (
    "Review: " + df["review"].astype(str) +
    " Aspect: " + df["aspect"].astype(str)
)

display(df[["bert_text", "sentiment_label"]].head(10))
```

```python
# STEP 8: Convert sentiment labels to numbers

label_to_id = {
    "Negative": 0,
    "Neutral": 1,
    "Positive": 2
}

id_to_label = {
    0: "Negative",
    1: "Neutral",
    2: "Positive"
}

df["label"] = df["sentiment_label"].map(label_to_id)

print(df[["bert_text", "sentiment_label", "label"]].head(10))
```

```python
# STEP 9: Train / Validation / Test split

from sklearn.model_selection import train_test_split

# First: 80% train, 20% temporary
train_df, temp_df = train_test_split(
    df,
    test_size=0.20,
    random_state=42,
    stratify=df["label"]
)

# Second: split temporary into 10% validation and 10% test
val_df, test_df = train_test_split(
    temp_df,
    test_size=0.50,
    random_state=42,
    stratify=temp_df["label"]
)

print("Training:", len(train_df))
print("Validation:", len(val_df))
print("Testing:", len(test_df))

print("\nTraining distribution:")
print(train_df["sentiment_label"].value_counts())

print("\nValidation distribution:")
print(val_df["sentiment_label"].value_counts())

print("\nTesting distribution:")
print(test_df["sentiment_label"].value_counts())
```

```python
import pandas as pd
import numpy as np

from datasets import Dataset
from transformers import (
    AutoTokenizer,
    AutoModelForSequenceClassification,
    TrainingArguments,
    Trainer
)

from sklearn.metrics import accuracy_score, precision_recall_fscore_support

print("Libraries loaded successfully!")
```

```python
train_data = Dataset.from_pandas(
    train_df[["bert_text", "label"]].reset_index(drop=True)
)

val_data = Dataset.from_pandas(
    val_df[["bert_text", "label"]].reset_index(drop=True)
)

test_data = Dataset.from_pandas(
    test_df[["bert_text", "label"]].reset_index(drop=True)
)

print("Train:", train_data)
print("Validation:", val_data)
print("Test:", test_data)
```

```python
model_name = "bert-base-uncased"

tokenizer = AutoTokenizer.from_pretrained(model_name)

print("Tokenizer loaded successfully!")
```

```python
def tokenize_function(examples):
    return tokenizer(
        examples["bert_text"],
        padding="max_length",
        truncation=True,
        max_length=128
    )

train_tokenized = train_data.map(
    tokenize_function,
    batched=True
)

val_tokenized = val_data.map(
    tokenize_function,
    batched=True
)

test_tokenized = test_data.map(
    tokenize_function,
    batched=True
)

print("Tokenization completed!")
```

```python
id_to_label = {
    0: "Negative",
    1: "Neutral",
    2: "Positive"
}

label_to_id = {
    "Negative": 0,
    "Neutral": 1,
    "Positive": 2
}

model = AutoModelForSequenceClassification.from_pretrained(
    model_name,
    num_labels=3,
    id2label=id_to_label,
    label2id=label_to_id
)

print("BERT model loaded!")
```

```python
def compute_metrics(eval_pred):
    logits, labels = eval_pred

    predictions = np.argmax(logits, axis=-1)

    precision, recall, f1, _ = precision_recall_fscore_support(
        labels,
        predictions,
        average="weighted",
        zero_division=0
    )

    accuracy = accuracy_score(labels, predictions)

    return {
        "accuracy": accuracy,
        "precision": precision,
        "recall": recall,
        "f1": f1
    }
```

```python
# STEP 17: Fixed BERT training configuration

from transformers import TrainingArguments

training_args = TrainingArguments(
    output_dir="/content/bert_automotive_model",

    eval_strategy="epoch",
    save_strategy="epoch",

    learning_rate=2e-5,

    per_device_train_batch_size=16,
    per_device_eval_batch_size=16,

    num_train_epochs=3,

    weight_decay=0.01,

    load_best_model_at_end=True,

    metric_for_best_model="f1",
    greater_is_better=True,

    report_to="none"
)

print("Training configuration ready!")
```

```python
import transformers

print("Transformers version:", transformers.__version__)
```

```python
# STEP 18: Create Trainer

trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=train_tokenized,
    eval_dataset=val_tokenized,
    processing_class=tokenizer,
    compute_metrics=compute_metrics
)

print("Trainer created successfully!")
```

```python
# FAST TRAINING VERSION

# Use only 5,000 training samples
fast_train_df = train_df.sample(
    n=min(5000, len(train_df)),
    random_state=42
)

# Use 1,000 validation samples
fast_val_df = val_df.sample(
    n=min(1000, len(val_df)),
    random_state=42
)

from datasets import Dataset

fast_train_data = Dataset.from_pandas(
    fast_train_df[["bert_text", "label"]].reset_index(drop=True)
)

fast_val_data = Dataset.from_pandas(
    fast_val_df[["bert_text", "label"]].reset_index(drop=True)
)

# Tokenize
fast_train_tokenized = fast_train_data.map(
    tokenize_function,
    batched=True
)

fast_val_tokenized = fast_val_data.map(
    tokenize_function,
    batched=True
)

print("Fast training dataset:", len(fast_train_tokenized))
print("Fast validation dataset:", len(fast_val_tokenized))
```

```python
training_args = TrainingArguments(
    output_dir="/content/bert_automotive_model",

    eval_strategy="epoch",
    save_strategy="no",

    learning_rate=2e-5,

    per_device_train_batch_size=32,
    per_device_eval_batch_size=32,

    num_train_epochs=1,

    weight_decay=0.01,

    report_to="none"
)

print("Fast training configuration ready!")
```

```python
trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=fast_train_tokenized,
    eval_dataset=fast_val_tokenized,
    processing_class=tokenizer,
    compute_metrics=compute_metrics
)

print("Fast Trainer ready!")
```

```python
trainer.train()
```

```python
# STEP 20: Evaluate BERT on the test dataset

test_results = trainer.evaluate(test_tokenized)

print("\n===== TEST RESULTS =====")

for key, value in test_results.items():
    if isinstance(value, float):
        print(f"{key}: {value:.4f}")
    else:
        print(f"{key}: {value}")
```

```python
# STEP 23: Test BERT with a new automotive review

def predict_sentiment(review, aspect):

    text = f"Review: {review} Aspect: {aspect}"

    inputs = tokenizer(
        text,
        return_tensors="pt",
        truncation=True,
        padding=True,
        max_length=128
    )

    # Move inputs to same device as model
    device = model.device
    inputs = {k: v.to(device) for k, v in inputs.items()}

    with torch.no_grad():
        outputs = model(**inputs)

    probabilities = torch.softmax(outputs.logits, dim=-1)
    prediction = torch.argmax(probabilities, dim=-1).item()

    return {
        "aspect": aspect,
        "sentiment": id_to_label[prediction],
        "confidence": probabilities[0][prediction].item()
    }
```
